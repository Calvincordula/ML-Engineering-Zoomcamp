```python
import pandas as pd

def extract_address_parts(address):
    country = ""
    city = ""
    zip_code = ""
    
    parts = address.split(',')
    if len(parts) >= 1:
        country = parts[-1].strip()
    if len(parts) >= 2:
        city = parts[-2].strip()
    if len(parts) >= 3:
        for part in parts:
            if any(c.isdigit() for c in part):
                zip_code = part.strip()
                break
    
    return country, city, zip_code

# Load the raw data and print column names to verify
data_path = "File1-2.xlsx"
raw_data = pd.read_excel(data_path, sheet_name=' Sheet1')

# Print column names to verify what we're working with
print("Available columns in the Excel file:")
print(raw_data.columns.tolist())

# Initialize lists for the two worksheets
original_data_rows = []
sll_spl_data_file_rows = []

# Process each entity in the raw data
for index, row in raw_data.iterrows():
    try:
        # Get values with exact column names
        entity_id = str(row['No. of the company'])  # Convert to string to preserve number format
        identification_data = str(row['Identification data (full name and details of the legal entity)']) if pd.notna(row['Identification data (full name and details of the legal entity)']) else ""
        sanction_type = str(row['type_of_preventive_measure_(in_accordance_with_the_law_of_ukraine_"on_sanctions")']) if pd.notna(row['type_of_preventive_measure_(in_accordance_with_the_law_of_ukraine_"on_sanctions")']) else ""
        period = str(row['Stagnation period']) if pd.notna(row['Stagnation period']) else ""
        
        if identification_data and identification_data.lower() != "nan":
            if 'Location: ' in identification_data:
                name_address_parts = identification_data.split('Location: ')
                entity_name = name_address_parts[0].strip()
                addresses = name_address_parts[1].split('; ') if len(name_address_parts) > 1 else []
            else:
                entity_name = identification_data
                addresses = []
            
            # Always add at least one row, even if no addresses
            if not addresses:
                original_data_rows.append([
                    entity_id,
                    entity_name,
                    sanction_type,
                    period
                ])
                
                sll_spl_data_file_rows.append([
                    entity_id,
                    entity_name,
                    "",  # Country
                    "",  # City
                    "",  # ZIP
                    sanction_type,
                    "",  # Valid From
                    ""   # Valid To
                ])
            else:
                # Add rows for each address
                for i, address in enumerate(addresses):
                    original_data_rows.append([
                        entity_id if i == 0 else '',
                        entity_name if i == 0 else '',
                        sanction_type if i == 0 else '',
                        period if i == 0 else ''
                    ])
                    
                    country, city, zip_code = extract_address_parts(address)
                    sll_spl_data_file_rows.append([
                        entity_id,
                        entity_name,
                        country,
                        city,
                        zip_code,
                        sanction_type,
                        "",  # Valid From
                        ""   # Valid To
                    ])
        else:
            # Handle cases where identification_data is missing
            original_data_rows.append([
                entity_id,
                "",  # Empty name
                sanction_type,
                period
            ])
            
    except Exception as e:
        print(f"Error processing row {index}: {str(e)}")
        continue

# Create DataFrames for the two sheets
original_data_df = pd.DataFrame(original_data_rows, columns=[
    '№', 
    'Identification Data (Full Name and Details of the Legal Entity)',
    'Type of Restrictive Measure (According to the Law of Ukraine "On Sanctions")',
    'Period of Application'
])

sll_spl_data_file_df = pd.DataFrame(sll_spl_data_file_rows, columns=[
    'SLL_SPL_DATA_REF_ID',
    'SLL_SPL_NAME',
    'SLL_SPL_COUNTRY',
    'SLL_SPL_CITY',
    'SLL_SPL_ZIP',
    'SLL_SPL_COMMENT',
    'SLL_SPL_VALID_FROM',
    'SLL_SPL_VALID_TO'
])

# Write the formatted data to a new Excel file
output_path = 'Formatted_Data.xlsx'
with pd.ExcelWriter(output_path) as writer:
    original_data_df.to_excel(writer, sheet_name='OriginalData', index=False)
    sll_spl_data_file_df.to_excel(writer, sheet_name='SLL_SPL_DATA_FILE', index=False)

print(f"Excel file has been created at: {output_path}")

# Print the first few rows of each DataFrame to verify the content
print("\nFirst few rows of OriginalData:")
print(original_data_df.head())
print("\nFirst few rows of SLL_SPL_DATA_FILE:")
print(sll_spl_data_file_df.head())
```
